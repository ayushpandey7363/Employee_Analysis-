```python
# Employee Salary Analysis System
# Dataset
import numpy as np

employee_id = np.array([
101,102,103,104,105,
106,107,108,109,110,
111,112,113,114,115,
116,117,118,119,120
])

department = np.array([
"IT","HR","Sales","IT","Finance",
"HR","Sales","IT","Marketing","Finance",
"IT","HR","Sales","Marketing","Finance",
"IT","Sales","HR","Finance","Marketing"
])

salary = np.array([
45000,35000,52000,61000,48000,
39000,55000,72000,43000,50000,
68000,36000,58000,47000,51000,
75000,62000,40000,54000,46000
])

experience = np.array([
2,1,4,6,3,
2,5,8,2,4,
7,1,6,3,5,
9,7,2,5,4
])

```

```python
# Total Employees
print("Total Employees:", len(employee_id))
```

```python
# Average Salary
print("Average Salary:", np.mean(salary))
```

```python
# Highest Salary
print("Highest Salary:", np.max(salary))
```

```python
# Lowest Salary
print("Lowest Salary:", np.min(salary))
```

```python
# Median Salary
print("Median Salary:", np.median(salary))
```

```python
# Standard Deviation
print("Standard Deviation:", np.std(salary))
```

```python

# Employees earning above ₹60,000
high_salary = salary[salary > 60000]

print(high_salary)
```

```python
# Employees earning below ₹40,000
print(salary[salary < 40000])
```

```python
# Give everyone a 10% salary increment
increment = salary * 1.10

print(increment)
```

```python
# Calculate 5% Tax
tax = salary * 0.05

print(tax)
```

```python
# Net Salary
net_salary = salary - tax

print(net_salary)
```

```python
# Find IT Department Employees
it_salary = salary[department == "IT"]

print(it_salary)
```

```python
# Employees with More Than 5 Years Experience
print(employee_id[experience > 5])
```
